## Ejercicio 1. Crear un árbol de categorías

Construya un árbol llamado `catalogo` con raíz **Tienda**, tres categorías (**Tecnología, Hogar, Libros**) y al menos dos productos en cada categoría. Muestre el árbol completo.

**Evidencia:** estructura visual con mínimo 10 nodos (1 raíz + 3 categorías + 6 productos).


```python
# Se utilizan identificadores breves y únicos; las etiquetas son los nombres visibles.
catalogo = Tree()
catalogo.create_node("Tienda", "tienda")

# Categorías principales
catalogo.create_node("Tecnología", "tecnologia", parent="tienda")
catalogo.create_node("Hogar", "hogar", parent="tienda")
catalogo.create_node("Libros", "libros", parent="tienda")

# Productos de Tecnología
catalogo.create_node("Computador portátil", "portatil", parent="tecnologia")
catalogo.create_node("Audífonos inalámbricos", "audifonos", parent="tecnologia")

# Productos de Hogar
catalogo.create_node("Licuadora", "licuadora", parent="hogar")
catalogo.create_node("Lámpara de mesa", "lampara", parent="hogar")

# Productos de Libros
catalogo.create_node("Cien años de soledad", "cien_anios", parent="libros")
catalogo.create_node("El principito", "principito", parent="libros")

print("ÁRBOL DEL CATÁLOGO")
catalogo.show()
print(f"Total construido: {catalogo.size()} nodos.")

# Verificación de los requisitos del enunciado
assert catalogo.root == "tienda"
assert catalogo.size() >= 10
assert all(len(catalogo.children(categoria)) >= 2 for categoria in ("tecnologia", "hogar", "libros"))
print("Verificación: raíz, tres categorías, mínimo dos productos por categoría y mínimo 10 nodos: OK.")
```

### 📝 Documentación del Ejercicio 1: Creación del Árbol de Categorías

* **Objetivo:** Construir una estructura jerárquica limpia con 10 nodos para simular un catálogo comercial.
* **Explicación del Código:**
  * **Instanciación:** Se crea un objeto vacío `Tree()` llamado `catalogo`.
  * **Nodo Raíz:** Se define el primer nodo con identificador `'tienda'`. Todos los demás nodos dependerán de este.
  * **Nodos Intermedios (Categorías):** Se agregan `'tecnologia'`, `'hogar'` y `'libros'` asignando como `parent="tienda"`.
  * **Nodos Hoja (Productos):** Se cuelgan 2 productos específicos bajo cada categoría correspondiente.
* **Concepto Teórico:** Un **árbol general** se diferencia de un árbol binario en que el grado de salida (número de hijos) de cualquier nodo puede ser mayor a 2, lo cual modela de forma perfecta taxonomías reales de clasificación.

### Análisis del resultado

La estructura tiene **tres niveles**: la raíz *Tienda*, las tres categorías y seis productos. Se obtienen exactamente 10 nodos, por lo que se cumple el mínimo solicitado sin agregar elementos ajenos al enunciado.

**Error frecuente y solución:** repetir un identificador genera un error aunque las etiquetas sean distintas. Por eso cada nodo usa un identificador único (`portatil`, `audifonos`, etc.) y el argumento `parent` siempre referencia un identificador ya creado.


## Ejercicio 2. Consultar relaciones

Utilice el árbol del ejercicio 1. Consulte: (a) padre de un producto, (b) hijos de una categoría, (c) hojas del árbol y (d) número total de nodos. Imprima resultados con etiquetas comprensibles.

**Evidencia:** salidas visibles y correctas.


```python
# (a) Padre de un producto
producto_consultado = "audifonos"
padre_producto = catalogo.parent(producto_consultado)
print(f"a) Padre de '{catalogo[producto_consultado].tag}': {padre_producto.tag}")

# (b) Hijos de una categoría
categoria_consultada = "tecnologia"
hijos_categoria = [nodo.tag for nodo in catalogo.children(categoria_consultada)]
print(f"b) Productos de '{catalogo[categoria_consultada].tag}': {hijos_categoria}")

# (c) Hojas del árbol
hojas = [nodo.tag for nodo in catalogo.leaves()]
print(f"c) Hojas del catálogo ({len(hojas)}): {hojas}")

# (d) Número total de nodos
total_nodos = catalogo.size()
print(f"d) Número total de nodos del catálogo: {total_nodos}")

# Verificación de coherencia
assert padre_producto.identifier == "tecnologia"
assert set(hijos_categoria) == {"Computador portátil", "Audífonos inalámbricos"}
assert len(hojas) == 6 and total_nodos == 10
print("Verificación de consultas: OK.")
```

### 📝 Documentación del Ejercicio 2: Consulta de Relaciones

* **Objetivo:** Extraer información estructural del árbol utilizando consultas de relación nativas de `treelib`.
* **Explicación del Código:**
  * **Relación de Paternidad:** El método `.parent(id)` devuelve el objeto nodo inmediatamente superior en la jerarquía.
  * **Relación de Descendencia:** El método `.children(id)` extrae un iterable de nodos que son hijos directos del identificador consultado.
  * **Identificación de Hojas:** El método `.leaves()` detecta todos aquellos nodos con grado de salida cero (es decir, nodos finales sin subcategorías ni productos asociados).
* **Concepto Teórico:** Las consultas de relaciones permiten navegar de manera bidireccional sobre la jerarquía, facilitando la extracción de métricas clave como el tamaño del árbol ($N=10$) sin necesidad de realizar un recorrido manual exhaustivo.

### Análisis del resultado

El padre de *Audífonos inalámbricos* es *Tecnología*, y los hijos consultados coinciden con los dos productos creados en esa categoría. Las seis hojas son productos porque no tienen descendientes; las categorías no son hojas y el total permanece en 10 nodos.

**Error frecuente y solución:** `parent()` y `children()` reciben el **identificador**, no la etiqueta visible. Las consultas usan `audifonos` y `tecnologia`, y convierten los nodos devueltos a etiquetas mediante `.tag` para producir mensajes comprensibles.


## Ejercicio 3. Modificar el árbol

Sobre una copia del árbol anterior, cambie el nombre visible de un producto, mueva un producto a otra categoría y elimine otro producto. Muestre el árbol antes y después.

**Pista:** revise `update_node()`, `move_node()` y `remove_node()` en la documentación.

**Evidencia:** comparación de las dos estructuras.


```python
# El constructor de Tree con deep=True crea una copia profunda e independiente.
catalogo_modificado = Tree(tree=catalogo, deep=True)

print("ANTES DE LAS MODIFICACIONES")
catalogo_modificado.show()

# 1) Cambiar únicamente la etiqueta visible del producto.
catalogo_modificado.update_node("portatil", tag="Portátil para estudio")

# 2) Mover la lámpara desde Hogar hacia Tecnología.
catalogo_modificado.move_node("lampara", "tecnologia")

# 3) Eliminar la licuadora.
catalogo_modificado.remove_node("licuadora")

print("DESPUÉS DE LAS MODIFICACIONES")
catalogo_modificado.show()

print("Cambio de nombre: 'Computador portátil' →", catalogo_modificado["portatil"].tag)
print("Nueva categoría de 'Lámpara de mesa':", catalogo_modificado.parent("lampara").tag)
print("¿Se eliminó 'Licuadora'?:", not catalogo_modificado.contains("licuadora"))
print("Nodos en la copia modificada:", catalogo_modificado.size())
print("Nodos en el árbol original:", catalogo.size())

# Verificación de las tres operaciones y de la conservación del original
assert catalogo_modificado["portatil"].tag == "Portátil para estudio"
assert catalogo_modificado.parent("lampara").identifier == "tecnologia"
assert not catalogo_modificado.contains("licuadora")
assert catalogo["portatil"].tag == "Computador portátil"
assert catalogo.parent("lampara").identifier == "hogar"
assert catalogo.contains("licuadora")
print("Verificación de update_node, move_node, remove_node y copia independiente: OK.")
```

### 📝 Documentación del Ejercicio 3: Mutación Jerárquica e Inmutabilidad

* **Objetivo:** Realizar de forma segura las tres modificaciones estructurales típicas en un árbol: edición de metadatos, reubicación de subramas y eliminación de elementos.
* **Explicación del Código:**
  * **Copia Profunda (`deep=True`):** Crea un duplicado del árbol en un espacio de memoria diferente. Esto evita el efecto de 'mutación colateral' donde cambios en la copia alteran al original.
  * **Actualización (`update_node`):** Modifica únicamente la etiqueta visual (`tag`) manteniendo intacto el identificador único para no romper referencias de otros métodos.
  * **Reubicación (`move_node`):** Desconecta un nodo (y toda su descendencia si la tuviese) de su padre anterior y lo conecta a un nuevo padre.
  * **Eliminación (`remove_node`):** Remueve por completo el nodo del sistema de indexación del árbol.

### Análisis del resultado

La comparación evidencia tres cambios independientes: `update_node()` modifica la etiqueta sin cambiar el identificador; `move_node()` conserva el producto, pero cambia su padre; y `remove_node()` reduce la copia de 10 a 9 nodos. El catálogo original sigue intacto, condición necesaria para utilizarlo en el ejercicio 4.

**Errores frecuentes y soluciones:** modificar `catalogo` directamente impediría comparar con la versión original, por lo que se usa `Tree(tree=catalogo, deep=True)`. Además, al mover un nodo se indican identificadores válidos y, antes de eliminar, se confirma que *Licuadora* es una hoja para evitar borrar accidentalmente una rama completa.


## Ejercicio 4. Recorridos

Sobre el árbol del ejercicio 1, liste los identificadores o etiquetas visitados en **profundidad (DFS)** y **anchura (BFS)**. Indique cuál recorrido visita primero todas las categorías.

**Pista:** `expand_tree(mode=Tree.DEPTH)` y `expand_tree(mode=Tree.WIDTH)`.

**Evidencia:** dos listas de recorrido y una conclusión breve (1–2 frases).


```python
recorrido_dfs = [catalogo.get_node(i).tag for i in catalogo.expand_tree(mode=Tree.DEPTH)]
recorrido_bfs = [catalogo.get_node(i).tag for i in catalogo.expand_tree(mode=Tree.WIDTH)]

print("DFS (profundidad):", recorrido_dfs)
print("BFS (anchura):", recorrido_bfs)
print("Conclusión: BFS visita primero todas las categorías, porque recorre por niveles. "
      "DFS completa la rama de una categoría antes de continuar con la siguiente.")

# En BFS, las tres categorías ocupan las posiciones 1, 2 y 3 después de la raíz.
assert set(recorrido_bfs[1:4]) == {"Tecnología", "Hogar", "Libros"}
assert recorrido_dfs[0] == recorrido_bfs[0] == "Tienda"
print("Verificación de recorridos sobre el árbol original: OK.")
```

### 📝 Documentación del Ejercicio 4: Análisis de Algoritmos de Recorrido

* **Objetivo:** Ilustrar la diferencia práctica de orden de visita entre los recorridos en profundidad y en anchura.
* **Explicación del Código:**
  * **DFS (Depth-First Search):** Explora de forma vertical. Entra a una categoría y desciende inmediatamente a sus productos antes de dar paso a las demás categorías.
  * **BFS (Breadth-First Search):** Explora de forma horizontal por niveles. Visita primero la raíz, luego todas las categorías hermanas del nivel 1, y posteriormente todos los productos del nivel 2.
* **Conclusión Práctica:** Para encontrar elementos agrupados u organizar una vista de menús de primer nivel de forma ágil, **BFS** resulta más eficiente.

### Análisis del resultado

Ambos recorridos comienzan en *Tienda*, pero siguen estrategias distintas. DFS desciende hasta los productos de una categoría antes de cambiar de rama; BFS procesa primero el nivel completo de categorías y después visita los productos.

**Error frecuente y solución:** `expand_tree()` produce identificadores. Para presentar etiquetas legibles, cada identificador se transforma con `catalogo.get_node(i).tag`; además, se emplean exactamente `Tree.DEPTH` y `Tree.WIDTH` para no confundir los modos.


## Ejercicio 5. Caso aplicado: servicios universitarios

Diseñe un árbol de navegación para un portal universitario con raíz **Portal**, al menos cuatro servicios principales y dos trámites por servicio.

Implemente una función `ruta_hasta_raiz(arbol, identificador)` que devuelva la ruta desde un trámite hasta la raíz (en orden raíz → trámite). Pruebe la función con dos trámites diferentes.

**Evidencia:** árbol visual y dos rutas completas.


```python
portal = Tree()
portal.create_node("Portal", "portal")

# Cuatro servicios principales
portal.create_node("Admisiones", "admisiones", parent="portal")
portal.create_node("Registro Académico", "registro", parent="portal")
portal.create_node("Bienestar Universitario", "bienestar", parent="portal")
portal.create_node("Biblioteca", "biblioteca", parent="portal")

# Dos trámites por servicio
portal.create_node("Solicitud de admisión", "solicitud_admision", parent="admisiones")
portal.create_node("Consulta de resultados", "resultados_admision", parent="admisiones")

portal.create_node("Inscripción de asignaturas", "inscripcion_asignaturas", parent="registro")
portal.create_node("Certificado de notas", "certificado_notas", parent="registro")

portal.create_node("Cita de orientación", "cita_orientacion", parent="bienestar")
portal.create_node("Inscripción a deportes", "inscripcion_deportes", parent="bienestar")

portal.create_node("Préstamo de libros", "prestamo_libros", parent="biblioteca")
portal.create_node("Renovación de préstamo", "renovacion_prestamo", parent="biblioteca")


def ruta_hasta_raiz(arbol, identificador):
    """Devuelve las etiquetas desde la raíz hasta el nodo solicitado."""
    if not arbol.contains(identificador):
        raise ValueError(f"No existe un nodo con identificador '{identificador}'.")

    ruta = []
    nodo_actual = arbol.get_node(identificador)
    while nodo_actual is not None:
        ruta.append(nodo_actual.tag)
        nodo_actual = arbol.parent(nodo_actual.identifier)

    return list(reversed(ruta))


print("ÁRBOL DE NAVEGACIÓN DEL PORTAL")
portal.show()

ruta_1 = ruta_hasta_raiz(portal, "solicitud_admision")
ruta_2 = ruta_hasta_raiz(portal, "prestamo_libros")
print("Ruta 1:", " → ".join(ruta_1))
print("Ruta 2:", " → ".join(ruta_2))

# Verificación de requisitos y rutas
servicios = portal.children("portal")
assert len(servicios) >= 4
assert all(len(portal.children(servicio.identifier)) >= 2 for servicio in servicios)
assert ruta_1 == ["Portal", "Admisiones", "Solicitud de admisión"]
assert ruta_2 == ["Portal", "Biblioteca", "Préstamo de libros"]
print("Verificación: cuatro servicios, dos trámites por servicio y dos rutas completas: OK.")
```

### 📝 Documentación del Ejercicio 5: Caso Aplicado e Historial de Navegación

* **Objetivo:** Resolver un problema real de navegación (Breadcrumbs / Migas de pan) mapeando la ruta desde cualquier servicio/trámite final hasta la página principal del portal.
* **Explicación de la Función `ruta_hasta_raiz`:**
  1. **Control de Errores:** Valida la existencia del nodo con `.contains()`. Si no existe, lanza un `ValueError` controlado.
  2. **Bucle de Ascenso:** Mediante un bucle `while` escalamos a través de los nodos padres (`arbol.parent()`) acumulando sus etiquetas.
  3. **Inversión de Orden:** Dado que el ascenso es de abajo hacia arriba (`Trámite → Raíz`), se aplica `reversed()` para que el usuario reciba la ruta en el orden de lectura natural (`Raíz → Servicio → Trámite`).

### Análisis del resultado

El portal contiene 13 nodos: una raíz, cuatro servicios y ocho trámites. La función asciende desde el nodo consultado mediante `parent()`, acumula las etiquetas y al final invierte la lista; así devuelve cada ruta en el orden exigido, desde *Portal* hasta el trámite.

**Error frecuente y solución:** si el identificador no existe, `get_node()` no permitiría construir una ruta válida. La función detecta ese caso con `contains()` y genera un `ValueError` descriptivo; para una consulta válida, el ciclo termina cuando el padre de la raíz es `None`, evitando accesos fuera del árbol.

### 💡 Conclusiones Clave

* **Unicidad e Identidad:** En `treelib`, la identidad de un nodo está estrictamente disociada de su representación visual. Utilizar identificadores cortos y únicos evita conflictos lógicos de duplicidad y permite escalar la estructura a cientos de nodos de forma limpia.
* **Profundidad frente a Anchura:**
  * **DFS** es el recorrido ideal cuando se requiere explorar ramas por completo antes de tomar decisiones o cuando la memoria es un factor restrictivo en árboles anchos.
  * **BFS** es el algoritmo óptimo para recorrer un árbol por niveles (jerarquías), lo que permite identificar rápidamente los elementos más cercanos a la raíz (como categorías globales).
* **La Importancia del Copiado Profundo:** Al modificar árboles, trabajar sobre copias superficiales (shallow copies) altera la fuente de datos original debido al paso de referencias de memoria en Python. La copia profunda (`deep=True`) garantiza la integridad e inmutabilidad de los datos históricos.
* **Ascenso Jerárquico:** El proceso de búsqueda desde un nodo hoja hasta la raíz es un algoritmo de costo lineal $O(h)$ (donde $h$ es la altura del árbol) sumamente eficiente, lo que facilita la creación de barras de navegación (breadcrumbs) en sistemas y aplicaciones web reales.
